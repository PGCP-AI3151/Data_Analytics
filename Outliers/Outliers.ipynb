```python
airline_df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Downloads\airline_passenger_satisfaction.csv")
airline_df
on_board_service = airline_df['on-board service']

```

```python
on_board_services = airline_df['On-board Service']
```

```python
mean_rating = on_board_services.mean()
median_rating = on_board_services.median()
skewness_rating = on_board_services.skew()
kurtosis_rating = on_board_services.kurt()
```

```python
print(on_board_service.value_counts().sort_index())
print(f"mean on board service: {mean_rating:.2f}")
print(f"median on board service: {median_rating:.2f}")
print(f"kurtosis of on board service: {kurtosis_rating:.2f}")
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x=on_board_service, color = 'skyblue')
plt.axvline(mean_rating, color='red', ls='--', label=f"mean: {mean_rating:.2f}")
plt.axvline(median_rating, color='black', ls='--', label=f"median: {median_rating:.2f}")
plt.title(f"Distribution of on_board service")
plt.legend()
plt.show()
```
