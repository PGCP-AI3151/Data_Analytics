```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('tips.csv')
df.head()
```

```python
sns.histplot(df['total_bill'],bins=10,color='darkgreen')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='gender', y='tip', data=df, hue = 'gender')
plt.title('Tip by Gender')
plt.show()
```

```python
sns.countplot(x='smoker', data=df, hue='smoker')
plt.title('count of smokers vs non smokers')
plt.show()
```

```python
sns.regplot(x='total_bill', y='tip', data=df, color='red')
plt.title('Total Bill vs Tip with Regression line')
plt.show()
```

```python
sns.barplot(x='day', y='tip', data=df)
plt.title('Average Tip by Day')
plt.show()
```
