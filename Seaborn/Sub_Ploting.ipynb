```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('tips.csv')
```

```python
#-----Functional Syntax-------
plt.figure(figsize=(12,5))
# Subplot 1
avg_tip_day = df.groupby('day')['tip'].mean()
plt.subplot(1,2,1)
plt.bar(avg_tip_day.index, avg_tip_day.values, color=['pink', 'lightblue', 'lightgreen', 'orange'],edgecolor ='black')
plt.title('Average Tip of Gender')
plt.ylabel('Average Tip')

#subplot 2

smoker_counts = df['smoker'].value_counts()
plt.subplot(1,2,2)
plt.pie(smoker_counts, labels=smoker_counts.index,autopct='%1.1f%%', colors=['gold','lightcoral'],shadow=True,
        wedgeprops = {'edgecolor' :'black'},
        explode = [0.03,0])
plt.title('Smoker vs Non-Smoker')
plt.tight_layout()
plt.show()
```

```python
# -------- OOP --------
fig, axes = plt.subplots(1,2,figsize=(12,5))
#subplot 1
axes[0].plot(df['total_bill'].iloc[:20], marker='o',linestyle='-',color='purple')
axes[0].set_title('Total Bill (First 20 Records)')
axes[0].set_xlabel('Record Index')
axes[0].set_ylabel('Total Bill')

#subplot 2
avg_tip_gender = df.groupby('gender')['tip'].mean()
axes[1].barh(avg_tip_gender.index, avg_tip_gender.values, color=['pink','lightblue'])
axes[1].set_title('Average Tip by Gender')
axes[1].set_xlabel('Average Tip')
plt.tight_layout()
plt.show()
```
