Independent Two-Sample t-Test\
Test difference in mean mpg between 4-cylinder and 6-cylinder cars\
Perform t-test and display results\
Conclusion on hypothesis

Expected output:\
Hypotheses and sample means\
t-test statistics and confidence interval\
p-value significance conclusion\
Interpretation statement

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Downloads\mtcars.csv")
df.head()
```

```python
mean_4cyl = df[df.cyl == 4]['mpg']
mean_6cyl = df[df.cyl == 6]['mpg']
print('4 Cylinder MPG Mean :-',mean_4cyl.mean().round(2))
print('6 Cylinder MPG Mean :-',mean_6cyl.mean().round(2))
```

```python
from scipy.stats import ttest_ind
result  = ttest_ind(mean_4cyl, mean_6cyl,alternative='two-sided')
print('T-Statistic :-',result.statistic)
print('P-Value :-',result.pvalue)
```

```python
alpha = 0.05
print("Null Hypothesis (H0)       : μ(4-cylinder) = μ(6-cylinder)")
print("Alternative Hypothesis(H1) : μ(4-cylinder) ≠ μ(6-cylinder)")

if result.pvalue < alpha:
    print('Reject H0 : Evidence that both mean are Different.')                  
else :
    print('Failed To Reject H0 : Evidence that both mean are Same.')
```

Interpretation Statement -> Above test we perform to check whether the mean of two independent group are 
                          significantly different. If p_value < alpha then we can reject H0 and say that the mean 
                            of these two independent group is different.

Q2

One-Way ANOVA Test

Test if mean hp differs by gear group\
Perform ANOVA and show summary\
Interpret results

Expected output:

Hypotheses stated\
ANOVA table with key statistics\
p-value significance conclusion\
Interpretation statement

```python
from scipy.stats import f_oneway
```

```python
mean_3gr = df[df['gear'] == 3]['hp']
mean_4gr = df[df['gear'] == 4]['hp']
mean_5gr = df[df['gear'] == 5]['hp']

print('3 Gear Hp Mean :-',mean_3gr.mean().round(2))
print('4 Gear Hp Mean :-',mean_4gr.mean().round(2))
print('5 Gear Hp Mean :-',mean_5gr.mean().round(2))
```

```python
result = f_oneway(mean_3gr,mean_4gr,mean_5gr)
print('f-statistic :-',result.statistic)
print('P-Value :-',result.pvalue)
```
