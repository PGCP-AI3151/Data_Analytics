Q9

Write a NumPy program to create an element-wise comparison (greater, greater_equal, less and
less_equal) of two given arrays.

```python
a1 = np.array([12,23,45,65])
a2 = np.array([67,98,56,34])

print(np.greater(a1,a2))
print(np.greater_equal(a1,a2))
print(np.less(a1,a2))
print(np.less_equal(a1,a2))
```
