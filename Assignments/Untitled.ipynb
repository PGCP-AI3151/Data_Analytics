```python
import pandas as pd
import numpy as np
df1= pd.Series([2, 4, 6, 8, 10])
df2 = pd.Series([1, 3, 5, 7, 9])
print(df1 + df2)
print(df1 * df2)
print(df1 / df2)
print(df1 - df2)
```

```python
df1 = pd.Series([2, 4, 6, 8, 10])
df2 = pd.Series([1, 3, 5, 7, 9])
print(df1.compare(df2))
print(df1.equals(df2))
print(df1==df2)
```

```python
dict = {'a': 100, 'b': 200, 'c': 300, 'd': 400, 'e': 800}
print('original:', dict)
df = pd.Series(dict)
print('converted series:', df)
```
