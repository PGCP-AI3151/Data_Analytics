```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

Q1

Write a Python program to draw a line with suitable label in the x axis, y
axis and a title.

```python
x_axis = np.array([0,1,2,3,4,5])
sns.lineplot(30 * x_axis,color='b')
plt.xticks(x_axis,x_axis*10)
plt.margins(x=0.01,y=0.01)
plt.title('Line Plot')
plt.xlabel('x-axis')
plt.ylabel('y-aix')
plt.tick_params(direction='in',top=True,right=True)
plt.show()
```

Q4

Write a Python program to draw line charts of the financial data of Alphabet
Inc. between October 3, 2016 to October 7, 2016.\
Sample Financial data (fdata.csv):\
Date,Open,High,Low,Close\
10-03-16,774.25,776.065002,769.5,772.559998\
10-04-16,776.030029,778.710022,772.890015,776.429993\
10-05-16,779.309998,782.070007,775.650024,776.469971\
10-06-16,779,780.47998,775.539978,776.859985\
10-07-16,779.659973,779.659973,770.75,775.080017

```python
data = {
    "Date": ["10-03-16", "10-04-16", "10-05-16", "10-06-16", "10-07-16"],
    "Open": [774.25, 776.030029, 779.309998, 779.0, 779.659973],
    "High": [776.065002, 778.710022, 782.070007, 780.47998, 779.659973],
    "Low": [769.5, 772.890015, 775.650024, 775.539978, 770.75],
    "Close": [772.559998, 776.429993, 776.469971, 776.859985, 775.080017]
}

df = pd.DataFrame(data)
y_axis = df.drop('Date',axis=1)
x_axis = df['Date']
```

```python
plt.plot(x_axis,y_axis,label=y_axis.columns)
plt.legend()
plt.xticks(x_axis.values,['3\nOct\n2016',4,5,6,7])
plt.show()
```

Q5

Write a Python program to plot two or more lines on same plot with suitable
legends of each line.

```python
sns.lineplot(x=[10,20,30],y=[20,40,10],color='b',label='line 1')
sns.lineplot(x=[10,20,30],y=[40,10,30],color='g',label='line 2')
plt.margins(x=0,y=0)
plt.xticks(np.linspace(10,30,5))
plt.tick_params(direction='in',top=True,right=True)
plt.legend()
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Two or More Lines On Same Plot With Suitable Legend')
plt.show()
```

Q6

Write a Python program to plot two or more lines with legends, different
widths and colors.

```python
sns.lineplot(x=[10,20,30],y=[20,40,10],color='b',label='line 1 - Width 3',linewidth=3)
sns.lineplot(x=[10,20,30],y=[40,10,30],color='r',label='line 2 - Width 5',linewidth=5)
plt.margins(x=0,y=0)
plt.xticks(np.linspace(10,30,5))
plt.tick_params(direction='in',top=True,right=True)
plt.legend()
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Two or More Lines With Different Width')
plt.show()
```

Q7

Write a Python program to plot two or more lines with different styles.

```python
sns.lineplot(x=[10,20,30],y=[20,40,10],color='b',label='line 1 - dotted',linewidth=2,linestyle='dotted')
sns.lineplot(x=[10,20,30],y=[40,10,30],color='r',label='line 2 - dashed',linewidth=4, linestyle='dashed')
plt.margins(x=0,y=0)
plt.xticks(np.linspace(10,30,5))
plt.tick_params(direction='in',top=True,right=True)
plt.legend()
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Two or More Lines With Different Style')
plt.show()
```

Q8

Write a Python program to plot two or more lines and set the line markers.

```python
sns.lineplot(x=[10,20,30],y=[20,40,10],color='r',linestyle='-.',marker='o',
                markerfacecolor="yellow", markeredgecolor="red",markersize=10 )
sns.lineplot(x=[10,20,30],y=[40,10,30],color='b',linestyle='--',marker='o',
                markerfacecolor="orange", markeredgecolor="blue",markersize=10 )
plt.xticks(np.linspace(10,30,5))
plt.tick_params(direction='in',top=True,right=True)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Display Marker')
plt.show()
```
