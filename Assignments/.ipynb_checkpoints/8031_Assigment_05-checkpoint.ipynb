Q49

Create a second DataFrame and concatenate it with the first one.

```python
data = {'Name':['Nitin','Om'],
        'Age':[23,19],
        'Marks':[89,91],
        'Result':['Pass','Pass'],
        'Bonus':[10,10],
        'Remark':['A','A']
       }
df2 = pd.DataFrame(data)
df2
```

```python
pd.concat([df,df2],ignore_index=True,join='inner')
```

Q50

Create two DataFrames with a common column and merge them.

```python
df1 = pd.DataFrame({
    'Name':['A','B','C','D','E'],
    'Math':[78,67,89,56,90]
})

df2 = pd.DataFrame({
    'Name':['A','B','C','D','X'],
    'Science':[91,80,66,86,93]
})

pd.merge(df1,df2,on='Name',how='inner')
```
