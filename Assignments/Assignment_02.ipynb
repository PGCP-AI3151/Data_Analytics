Write a Pandas program to compute the Euclidean distance between two given series.
Euclidean distance
From Wikipedia,
In mathematics, the Euclidean distance or Euclidean metric is the "ordinary" straight-line

Q32

Write a Pandas program to find the positions of the values neighboured by smaller
values on both sides in a given series.

```python
df = pd.Series([1,8,7,5,6,5,3,4,7,1])
temp = np.diff(np.sign(np.diff(df)))
result = np.where(temp == -2)[0] + 1
result
```

Autocorrelation, also known as serial correlation, is the correlation of a signal with a
