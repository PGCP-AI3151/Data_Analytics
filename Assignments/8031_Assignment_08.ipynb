Q1. Write a Pandas program to create

a) Datetime object for Jan 15 2012.\
b) Specific date and time of 9:20 pm.\
c) Local date and time.\
d) A date without time.\
e) Current date.\
f) Time from a datetime.\
g) Current local time.

```python
import pandas as pd
import datetime
```

```python
date = pd.to_datetime('Jan 15 2012')
print('Year :-',date.year)
print('Month :-',date.month)
print('Day :-',date.day)
print('Day Of Week :-',date.day_name())
```

```python
ts = pd.Timestamp('Jan 15 2012 21:20:00')
print('Specific date and time of 9:20 pm :-',ts)
```

```python
print('Local Date & Time :-',datetime.datetime.now())
print('Local Date & Time :-',pd.Timestamp.now())
```

```python
print('----- Using DateTime -----')
print('Current Date :-',datetime.datetime.now().date())
print('Current Time :-',datetime.datetime.now().time())

print('\n----- Using Timestamp -----')
print('Current Date :-',pd.Timestamp.now().date())
print('Current Time :-',pd.Timestamp.now().time())
```

```python
print('Time From a Datetime :-',ts.time().strftime("%I:%M:%S %p"))
```

```python
print('Current Local Time :-',pd.Timestamp.now().time().strftime('%I:%M:%S %p'))
```

Q4

Write a Pandas program to print the day after and before a specified date. Also
print the days between two given dates.

```python
date = pd.Timestamp('2026-10-30')
print('Specified Date :-',date.date(),' Day :-',date.day_name())

before = date - pd.Timedelta(days=1)
print('Before Date :-',before.date(),' Day :-',before.day_name())

after = date + pd.Timedelta(days=1)
print('After Date :-',after.date(),' Day :-',after.day_name())

start_date = pd.Timestamp("2026-10-30")
end_date = pd.Timestamp("2026-11-15")
duration = end_date - start_date
print(f'Days Between {start_date.date()} & {end_date.date()} :- {duration.days}')
```

Q5

Write a Pandas program to create a time-series with two index labels and
random values. Also print the type of the index.

```python
time_series = pd.date_range('2026-10-30',periods=3,freq='D')
data = {'Sale' :[100,120,90,125,110,60]}

stores = ['Store A','Store B']

multi_index = pd.MultiIndex.from_product([time_series,stores],names=['Date','Series'])
df = pd.DataFrame(data,index=multi_index)
print(f'Index Type :-{type(df.index)}')
df
```

Q7

data. Also select the dates of same year and select the dates between certain
dates.

```python
dates = ['2026-10-11','2026-10-15','2026-10-12','2026-10-13','2026-10-14']
ts = pd.to_datetime(dates)
data = [10,20,30,40,50]

time_series = pd.Series(data, index=ts,name='Time-Series')
time_series.sort_index(inplace=True)
time_series.loc['2026-10-11':'2026-10-14']
```

Q15

Write a Pandas program to get a time series with the last working days of
each month of a specific year.

```python
year = 2026
last_working_days = pd.date_range(start=f'{year}-01-01', end=f'{year}-12-31', freq='BME')
df = pd.DataFrame({
    'Last Working Day': last_working_days,
    'Day of Week': last_working_days.day_name()
})
df
```

Q16

Write a Pandas program to create a time series combining hour and minute.

```python
  df = pd.DataFrame({
    'hour':,
    'minute': [0, 30, 45, 15]
})

base_date = pd.to_datetime('2026-10-13')
df['timestamp'] = (
    base_date 
    + pd.to_timedelta(df['hour'], unit='H') 
    + pd.to_timedelta(df['minute'], unit='T')
)
df
```
