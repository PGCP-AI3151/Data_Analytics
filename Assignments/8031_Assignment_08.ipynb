Q1. Write a Pandas program to create

a) Datetime object for Jan 15 2012.\
b) Specific date and time of 9:20 pm.\
c) Local date and time.\
d) A date without time.\
e) Current date.\
f) Time from a datetime.\
g) Current local time.

```python
import pandas as pd
import datetime
```

```python
date = pd.to_datetime('Jan 15 2012')
print('Year :-',date.year)
print('Month :-',date.month)
print('Day :-',date.day)
print('Day Of Week :-',date.day_name())
```

```python
ts = pd.Timestamp('Jan 15 2012 21:20:00')
print('Specific date and time of 9:20 pm :-',ts)
```

```python
print('Local Date & Time :-',datetime.datetime.now())
print('Local Date & Time :-',pd.Timestamp.now())
```

```python
print('----- Using DateTime -----')
print('Current Date :-',datetime.datetime.now().date())
print('Current Time :-',datetime.datetime.now().time())

print('\n----- Using Timestamp -----')
print('Current Date :-',pd.Timestamp.now().date())
print('Current Time :-',pd.Timestamp.now().time())
```

```python
print('Time From a Datetime :-',ts.time().strftime("%I:%M:%S %p"))
```

```python
print('Current Local Time :-',pd.Timestamp.now().time().strftime('%I:%M:%S %p'))
```
