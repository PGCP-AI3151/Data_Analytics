Q20

Write a Pandas program to create a hitmap for more information about the

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.heatmap(df.isnull(),cbar=False,annot=True,cmap='Blues',linewidths=1, linecolor='black')
plt.title('Distribution of Missing Value Accross Columns',fontweight='bold')
plt.xlabel('Columns',fontweight='bold')
plt.ylabel('Row Index',fontweight='bold')
plt.show()
```
