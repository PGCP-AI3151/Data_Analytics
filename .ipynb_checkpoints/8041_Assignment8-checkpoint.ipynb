### **Question 1**
Write a Pandas program to create:
a) Datetime object for Jan 15 2012.  
b) Specific date and time of 9:20 pm.  
c) Local date and time.  
d) A date without time.  
e) Current date.  
f) Time from a datetime.  
g) Current local time.

```python
import pandas as pd
import datetime

# a)
print("a) Datetime object for Jan 15 2012:")
print(pd.to_datetime('2012-01-15'))

# b)
print("\nb) Specific date and time of 9:20 pm:")
print(pd.to_datetime('2012-01-15 21:20:00'))

# c)
print("\nc) Local date and time:")
print(datetime.datetime.now())

# d)
print("\nd) A date without time:")
print(datetime.date.today())

# e)
print("\ne) Current date:")
print(pd.to_datetime('today').date())

# f)
print("\nf) Time from a datetime:")
print(datetime.datetime.now().time())

# g)
print("\ng) Current local time:")
print(pd.to_datetime('now').time())
```

### **Question 2**
Write a Pandas program to create:
a) a specific date using timestamp.  
b) date and time using timestamp.  
c) a time adds in the current local date using timestamp.  
d) current date and time using timestamp.

```python
# a)
print("a) Specific date using timestamp:")
print(pd.Timestamp('2023-10-01'))

# b)
print("\nb) Date and time using timestamp:")
print(pd.Timestamp('2023-10-01 14:30:15'))

# c)
print("\nc) Time added to current local date using timestamp:")
print(pd.Timestamp(datetime.date.today()) + pd.Timedelta(hours=5, minutes=30))

# d)
print("\nd) Current date and time using timestamp:")
print(pd.Timestamp.now())
```

### **Question 4**
Write a Pandas program to print the day after and before a specified date. Also print the days between two given dates.

```python
date = pd.to_datetime('2023-05-15')
print("Specified Date:", date)
print("Day Before:", date - pd.Timedelta(days=1))
print("Day After:", date + pd.Timedelta(days=1))

date1 = pd.to_datetime('2023-05-10')
date2 = pd.to_datetime('2023-05-15')
print("\nDays between:", (date2 - date1).days, "days")
```

### **Question 7**
Write a Pandas program to create a time series object that has time indexed data. Also select the dates of same year and select the dates between certain dates.

```python
dates = pd.date_range('2021-01-01', '2023-12-31', freq='M')
ts = pd.Series(range(len(dates)), index=dates)

print("Dates of the same year (2022):")
print(ts['2022'])

print("\nDates between '2021-06-01' and '2022-03-01':")
print(ts['2021-06-01':'2022-03-01'])
```

### **Question 12**
Write a Pandas program to convert year and day of year into a single datetime column of a dataframe.

```python
df = pd.DataFrame({'year': [2020, 2021, 2022], 'day_of_year': [100, 150, 200]})
df['datetime'] = pd.to_datetime(df['year'].astype(str) + '-' + df['day_of_year'].astype(str), format='%Y-%j')
print(df)
```

### **Question 20**
Write a Pandas program to subtract two timestamps of same time zone or different time zone.

```python
t1 = pd.Timestamp('2023-01-01 12:00:00', tz='UTC')
t2 = pd.Timestamp('2023-01-02 12:00:00', tz='UTC')
print("Same Timezone difference:", t2 - t1)

t3 = pd.Timestamp('2023-01-02 12:00:00', tz='US/Eastern')
print("Different Timezone difference:", t2 - t3.tz_convert('UTC'))
```

### **Question 25**
Write a Pandas program to extract the day name from a specified date. Add 2 days and 1 business day with the specified date.

```python
dt = pd.to_datetime('2023-05-19') # Friday
print("Day Name:", dt.day_name())
print("Add 2 days:", dt + pd.Timedelta(days=2))
print("Add 1 business day:", dt + pd.offsets.BusinessDay(1))
```

### **Question 27**
Write a Pandas program to calculate one, two, three business day(s) from a specified date. Also find the next business month end from a specific date.

```python
dt = pd.to_datetime('2023-05-19')
for i in [1, 2, 3]:
    print(f"{i} business day(s) from {dt.date()}:", (dt + pd.offsets.BusinessDay(i)).date())
print("Next business month end:", (dt + pd.offsets.BMonthEnd(1)).date())
```

### **Question 28**
Write a Pandas program to create a period index represent all monthly boundaries of a given year. Also print start and end time for each period object in the said index.

```python
p_idx = pd.period_range(start='2023-01', end='2023-12', freq='M')
for p in p_idx:
    print(f"Period: {p} | Start: {p.start_time} | End: {p.end_time}")
```

### **Question 29**

### **Question 30**
