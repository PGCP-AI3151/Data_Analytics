```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('tips.csv')
df.head()
```

```python
df['tip_percentage'] = (100 * df['tip']/df['total_bill']).round(2)
```

```python
df['price_per_person'] = (df['total_bill']/df['size']).round(2)
```

```python
df[['total_bill', 'tip', 'size', 'tip_percentage', 'price_per_person']].head()
```

```python
# Condition Column Creation
```

```python
df['big_tipper'] = np.where(df['tip_percentage'] > 20 ,'Yes','No')
```

```python
# Several Outcomes : np.select(conditions,choices,default)
```

```python
df['bill_category'] = np.select([df['total_bill'] < 10 , df['total_bill'] <30],['$','$$'],default='$$$')
```

```python
df[['total_bill','bill_category']]
```

```python
# rename and drop column
```

```python
df = df.rename(columns={'total_bill':'bill','Payer Name':'payer_name'})
```

```python
df = df.drop(columns = ['CC Number', 'Payment ID'])
```

```python
# data type conversion
```

```python
df['size'] = df['size'].astype(float)  #int->float
```

```python
df['day'] = df['day'].astype('category')
print(df.dtypes)
```
