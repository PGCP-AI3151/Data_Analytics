```python
import pandas as pd
import numpy as np
```

```python
raw = pd.Series(['10','25','abc',None,'7.5'])
print(raw)
print(pd.to_numeric(raw, errors = 'coerce'))    # bad values -> Nan
```

```python
print(pd.to_numeric(raw, errors = 'coerce').fillna(0).astype(int))  # bad values -> Nan -> 0 -> Convert to int
```

```python
# Boolean Filtering
```

```python
df = pd.read_csv('tips.csv')
df.head()
```

```python
print(df['total_bill'] > 30)  # boolean series (True,False)
```

```python
print(df[df['total_bill'] > 30])   # row where it is True
```

```python
# Multiple Condition
```

```python
print(df[(df['total_bill'] > 30) & (df['gender'] == 'Male')])
```

```python
print(df[(df['total_bill'] > 30) & ~(df['gender'] == 'Male')])
```

```python
# print(df[(df['total_bill'] > 30) & (df['gender'] != 'Male')])  -> same as above
```

```python
print(df[(df['total_bill'] > 30) |(df['tip'] > 5)])
```

```python
# Membership and range
```

```python
df[df['day'].isin(['Sat','Sun'])].shape
```

```python
df[df['total_bill'].between(10,20)].shape  # 10 <= bill <= 20 (both include)
```

```python
df[df['total_bill'].between(10,20,inclusive='left')].shape  # 10 <= bill < 20 (First include)
```

```python
# Text Column
```

```python
df[df['Payer Name'].str.contains('Smith',case=False)]
```

```python
df[df['Payment ID'].str.startswith('Sun')]
```

```python
# Same condition , 4 syntaxes
```

```python
filtered1 = df[(df['gender'] == 'Female') & (df['tip'] >=4)]
filtered1.head()
```

```python
min_tip = 4
filtered2 = df.query("gender == 'Female' and tip >= @min_tip")
filtered2.head()
```

```python
filtered3 = df.loc[(df['gender'] == 'Female') & (df['tip'] >= 4),['total_bill','tip','day']]
filtered3.head()
```

```python
df['tip'].where(df['tip'] >= 4)
```
