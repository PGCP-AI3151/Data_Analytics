```python
import pandas as pd
import numpy as np
```

```python
a = np.array([10,20,30])
b = np.array([1,2,3])
print(a+b)  #by position,no labels
```

```python
s1 = pd.Series([10,20,30], index = ['alice', 'bob', 'charlies'])
s2 = pd.Series([1,2,3,4], index = ['bob', 'charlies', 'alice', 'dave'])
print(s1 + s2)
```

```python
name = ['alice', 'bob', 'charlie']
marks = [67,36,81]
print(pd.Series(data=marks))
student = pd.Series(data=marks, index=name, name='marks')
print(student)
```

```python
np.random.seed(42)
ages = pd.Series(np.random.randint(18,60,4), index=['alice', 'bob', 'charlies', 'dave'])
print(ages)
```

```python
print(pd.Series({'alice': 21, 'bob': 26, 'charlie': 23}))
```

```python
q1 = {'japan': 80, 'china': 450, 'india': 200, 'usa': 250}
q2 = {'brazil': 100, 'china': 500, 'india': 210, 'usa': 260}
sales_q1 = pd.Series(q1, name= 'q1')
sales_q2 = pd.Series(q2, name= 'q2')
print(sales_q1)
```

```python
print(sales_q1.loc['japan'])
print(sales_q1.iloc[0])
```

```python
print(sales_q1.loc['china':'usa'])
print(sales_q1.iloc[1:3])
print(sales_q1.loc['india': 'usa'])
```

```python
print(sales_q1.index)
print(sales_q1.values)
print(sales_q1 * 2)
print(sales_q1 / 100)
print(sales_q1[sales_q1 > 150])
print(sales_q1.sum(), sales_q1.mean(), sales_q1.max())
print(sales_q1.idxmax())
print(sales_q1.sort_values(ascending = False))
```
