Q3

Write a Pandas program to add, subtract, multiple and divide two Pandas Series.
Sample Series: [2, 4, 6, 8, 10], [1, 3, 5, 7, 9]

```python
df1 = pd.Series([2, 4, 6, 8, 10])
df2 = pd.Series([1, 3, 5, 7, 9])

print(df1 + df2)
print(df1 - df2)
print(df1 * df2)
print(df1 / df2)
```

Q4

Write a Pandas program to compare the elements of the two Pandas Series.
Sample Series: [2, 4, 6, 8, 10], [1, 3, 5, 7, 10]

```python
df1 = pd.Series([2, 4, 6, 8, 10])
df2 = pd.Series([1, 3, 5, 7, 10])

print(df1 == df2)
print(df1.compare(df2))
print(df1.equals(df2))
```

Q16 

Write a Pandas program to get the items of a given series not present in another given
series.

```python
sr1 = pd.Series([1,2,0,7,8,5,3])
sr2 = pd.Series([5,7,8,6])

sr1[~sr1.isin(sr2)]
```
