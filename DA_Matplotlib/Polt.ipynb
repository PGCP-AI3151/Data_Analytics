```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('tips.csv')
df
```

```python
df[df['gender'] == 'Male'][['tip','day']]
```

```python
df.loc[df['gender'] == 'Male',['tip','day']]
```

```python
plt.figure(figsize=(8,4))
plt.hist(df['total_bill'], bins = 5, color='skyblue', edgecolor='black')
plt.title('Distribution of Total Bill')
plt.xlabel('Total bill')
plt.ylabel('Frequency')
plt.show()
```

```python
#  Bar Plot
```

```python
avg_tip_gender = df.groupby('gender')['tip'].mean()
print(avg_tip_gender)
plt.figure(figsize=(6,4))
plt.bar(avg_tip_gender.index, avg_tip_gender.values, color=['pink', 'lightblue'])
plt.title('Average Tip of Gender')
plt.ylabel('Average Tip')
plt.show()
```

```python
# Line Plot
```

```python
plt.figure(figsize=(6,4))
plt.plot(df.index, df['total_bill'], marker='o', linestyle = '-', color = 'purple')
plt.title('Total Bill Across Records')
plt.xlabel('Index')
plt.ylabel('Total Bill')
plt.show()
```

```python
# Scatter Plot
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['total_bill'], df['tip'], marker='o', linestyle = '-', color = 'orange',edgecolor = 'black')
plt.title('Tip Across Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```
