```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Q1

Write a Python programming to display a bar chart of the popularity of
programming Languages.
Sample data:\
Programming languages: Java, Python, PHP, JavaScript, C#, C++\
Popularity: 22.2, 17.6, 8.8, 8, 7.7, 6.7

```python
df = pd.DataFrame(
    {'Language':['Java','Python','PHP','JavaScript','C#','C++'],
     'Popularity':[22.2,17.6,8.8,8,7.7,6.7]
    })
df
```

```python
plt.bar(df.Language,df.Popularity,align='edge',width=0.7,edgecolor ='black',color='blue',alpha=0.9)
plt.margins(x=0)
plt.title('Popularity Of Programming Language \nWorldwide, Oct 2017 Compare to Year Ago')
plt.xlabel('Programming Languages' )
plt.ylabel('Popularity')
ax = plt.gca()
ax.minorticks_on()
ax.grid(which='major', linestyle='-', alpha=0.7)
ax.grid(which='minor', linestyle=':', alpha=0.3)
ax.set_axisbelow(True)
plt.show()
```

Q12

Write a Python program to create bar plots with error bars on the same
figure.
Sample Date\
Mean velocity: 0.2474, 0.1235, 0.1737, 0.1824\
Standard deviation of velocity: 0.3314, 0.2278, 0.2836, 0.2645

```python
df = pd.DataFrame({
    'Mean_Velocity':[0.2474,0.1235,0.1737,0.1824],
    'Sd_Velocity':[0.3314,0.2278,0.2836,0.2645]
})
df
```

```python
x = np.arange(len(df))
errors =  df.Sd_Velocity - df.Mean_Velocity 
plt.bar(x, df.Mean_Velocity ,color='red',width=0.5,edgecolor='black',yerr=errors,capsize=5)
plt.title('Mean Velocity Vs Standard Deviation Velocity')
plt.xticks(x,df.Sd_Velocity)
plt.xlabel('Standard Deviation Velocity')
plt.ylabel('Mean Velocity')
plt.margins(x=0)
ax = plt.gca()
ax.minorticks_on()
ax.grid(which='major', linestyle='-', alpha=0.7)
ax.grid(which='minor', linestyle=':', alpha=0.3)
ax.set_axisbelow(True)
plt.show()
```

Q16

Write a Python program to create stack bar plot and add label to each
section.
Sample data:\
people = ('G1','G2','G3','G4','G5','G6','G7','G8')\
segments = 4\
multi-dimensional data\
data = [[ 3.40022085, 7.70632498, 6.4097905, 10.51648577, 7.5330039,
7.1123587, 12.77792868, 3.44773477],
[ 11.24811149, 5.03778215, 6.65808464, 12.32220677, 7.45964195,
6.79685302, 7.24578743, 3.69371847],
[ 3.94253354, 4.74763549, 11.73529246, 4.6465543, 12.9952182,
4.63832778, 11.16849999, 8.56883433],
[ 4.24409799, 12.71746612, 11.3772169, 9.00514257, 10.47084185,
10.97567589, 3.98287652, 8.80552122]]

```python
people = ['G1', 'G2', 'G3', 'G4', 'G5', 'G6', 'G7', 'G8']
data = np.array([
    [3.40022085, 7.70632498, 6.4097905, 10.51648577, 7.5330039, 7.1123587, 12.77792868, 3.44773477],
    [11.24811149, 5.03778215, 6.65808464, 12.32220677, 7.45964195, 6.79685302, 7.24578743, 3.69371847],
    [3.94253354, 4.74763549, 11.73529246, 4.6465543, 12.9952182, 4.63832778, 11.16849999, 8.56883433],
    [4.24409799, 12.71746612, 11.3772169, 9.00514257, 10.47084185, 10.97567589, 3.98287652, 8.80552122]
])
colors = ['red', 'green', 'white', 'purple']

fig, ax = plt.subplots(figsize=(12, 6), layout="tight")
left_margin = np.zeros(len(people))

for i in range(len(data)):
    segment_values = data[i]
    bars = ax.barh(
        people, 
        segment_values, 
        left=left_margin, 
        color=colors[i], 
        edgecolor='black'
    )
    ax.bar_label(bars, label_type='center', fmt='%.1f%%', fontsize=10, 
                 color='black' if colors[i] in ['white', 'green'] else 'white')
    left_margin += segment_values

ax.set_xlabel('Scores')
ax.set_ylabel('Groups')
ax.set_title('Stacked Bar Plot with Segment Labels', pad=15)
plt.show()
```
