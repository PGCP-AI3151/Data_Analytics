```python
import numpy as np
```

```python
def roll_dice():
    # Simulation of rolling a dice twice, minimum number we get is 1, max is 6
    # Add a results of the two simulations, so possible values will be (1,1) or (1,2) .. (6,6).. sum will be between 2 and 12
    # Run multiple times to verify
    return np.sum(np.random.randint(1,7,2))

print(roll_dice())
print(roll_dice())
print(roll_dice())
```

```python
# Part 2 : Someone approaches us saying I Will give you 5 dollars if you get 7 and take 1 dollar if you get a number other than 7
# How do we know what will happen?
# Our own 'Monte_Carlo_Simulation' like function

def monte_carlo_simulation(runs=1000):
    result = np.zeros(2)  # win counts and loss count
    for _ in range(runs):
        if roll_dice() == 7:
            result[0] += 1
        else:
            result[1] += 1
    # print(f'On 1000 Trials Profit(+) or loss(-) -> {result[0] *5 - result[1]}')
    return result
# Test 2-3 times
print(monte_carlo_simulation())
print(monte_carlo_simulation())
print(monte_carlo_simulation())
```

```python
# Part 3 : Now do it 1000 times... takes some time 
results = np.zeros(1000)

for i in range(1000):
    results[i] = monte_carlo_simulation()[0]
print(results)
```

```python
# Let plot it 
import matplotlib.pyplot as plt
fig,ax = plt.subplots()
ax.hist(results,bins=15,edgecolor='black')
plt.show()
```

```python
# Our win/loss
print(results.mean()) # mean wins
print(results.mean() * 5) # winning amounts
```
