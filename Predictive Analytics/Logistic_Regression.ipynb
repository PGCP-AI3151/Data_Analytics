```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix,ConfusionMatrixDisplay
from sklearn.preprocessing import  StandardScaler
from sklearn.model_selection import train_test_split
import numpy as np
```

```python
## 1. Load and prepare dataset
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Data_Analytics\Analysis\diabetes.csv")
X = df.drop('Outcome',axis=1)
y = df['Outcome']
```

```python
## 2. Train-Test Split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=101,stratify=y)
```

```python
## 3. Standardize Insulin (fit on train, transform both)
scaler = StandardScaler()
X_train['Insulin'] = scaler.fit_transform(X_train[['Insulin']])
X_test['Insulin'] = scaler.transform(X_test[['Insulin']])
```

```python
## 4. Train Full Model (All Features)
model = LogisticRegression(max_iter=500)
model.fit(X_train,y_train)
```

```python
## 5. Evaluation On Test
y_pred = model.predict(X_test)
```

```python
cm = confusion_matrix(y_test,y_pred)
ConfusionMatrixDisplay(cm,display_labels=['No Diabetes','Diabetes']).plot(cmap='Blues')
plt.title('Confusion Matrix (Test Set)')
plt.show()
```

```python
## 6. Logistic Regression Curve (Glucose Only -for visalization)
# Extract glucose from train/test set
X_train_glu = X_train[['Glucose']].copy()
X_test_glu = X_test[['Glucose']].copy()
model_glucose = LogisticRegression(max_iter = 500)
model_glucose.fit(X_train_glu,y_train)
```

```python
# Train Separate single-featuree model for visualization
#create smooth curve
glu_min = df['Glucose'].min()
glu_max = df['Glucose'].max()
raw_values = np.linspace(glu_min,glu_max,200).reshape(-1,1)
glucose_values = pd.DataFrame(raw_values, columns=['Glucose'])
y_curve = model_glucose.predict_proba(glucose_values)[:,1]
```

```python
# Plot Sigmoid curve with train/test split visualization
plt.figure(figsize=(8,5))
plt.plot(glucose_values,y_curve,color='green',linewidth=2,label='Logistic Curve')
plt.scatter(X_train_glu,y_train,color='blue',alpha=0.4,label='Total Data')
plt.scatter(X_test_glu,y_test,color='red',alpha=0.6,label='Test Data',marker='x',s=80)
plt.xlabel('Glucose Level')
plt.ylabel('Probability Of Diabetes (outcome = 1)')
plt.title('Logistic Regression Curve - Glucose Feature Only')
plt.legend()
plt.grid(True,alpha=0.3)
plt.show()
```
