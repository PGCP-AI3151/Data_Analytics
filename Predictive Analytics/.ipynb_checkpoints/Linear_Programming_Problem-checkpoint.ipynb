```python
# Linear Programming Problem (LLP) : how many chairs and tables should a factory make to maximise profit ?
# ! pip install pulp==3.3.2
```

```python
import numpy as np
from pulp import LpInteger, LpMaximize, LpProblem, LpVariable, value
```

```python
model = LpProblem('FactoryProductionProblem',LpMaximize)
```

```python
# Parameter : name , lowe limit, upper limit ,type
chairs = LpVariable('Chairs',lowBound=0,upBound=None,cat=LpInteger)
tables = LpVariable('Tables',lowBound=0,upBound=None,cat=LpInteger)
```

```python
model += 300 * chairs + 1000 * tables,'profit'
# objective function
model += 5 * chairs + 10 * tables <= 400,'Wood_Constraint'
model += 2 * chairs + 6 * tables <= 300,'Time_Constraint'
model += chairs >= 2, 'Chairs_Constraint'
```

```python
model.solve()
for v in model.variables():
    print(v.name,' = ',v.varValue)
print('Total Profit :',value(model.objective))
```

# Excericise

```python
model = LpProblem('CarProfitProble',LpMaximize)
```

```python
carA = LpVariable('Car_A',lowBound=0,upBound=None,cat=LpInteger)
carB = LpVariable('Car_B',lowBound=0,upBound=None,cat=LpInteger)
```

```python
model += 30000 * carA + 45000 * carB,'Profit'
model += 3 * carA + 4 * carB <= 30 ,'Robot Time'
model += 5 * carA + 6 * carB <= 60 ,'Engineer Time'
model += 1.5 * carA + 3 * carB <= 21 ,'Derailer Time'
```

```python
model.solve()
for v in model.variables():
    print(v.name,' = ',v.varValue)
print('Total Profit :',value(model.objective))
```
