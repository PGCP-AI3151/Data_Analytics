```python
#1-sample T-test: is average population tip = $3?
#HO: meu = average tip=$3
import pandas as pd
from scipy import stats
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Data_Analytics\DA_Pandas\tips.csv")
df.head()
```

```python
our_column = df['tip']
our_column = our_column.sample(n=30,replace=False)
```

# Two-Tailed t_test

```python
meu = 3 #meu
```

```python
#perform one sample T-test
t_statistics, p_value = stats.ttest_1samp(our_column, meu)
```

```python
#automatically retuens 2-tailed p-value
print('p-value:', p_value)   
```

```python
alpha = 0.05
if p_value < alpha:
    print('Reject HO')
else :
    print('Failed To Reject Ho')
```

```python
our_column.mean()
```

# One-Tailed t_test

```python
# Perform one-sample right tailed test
# HO : mu <= 3
# Ha : mu > 3

result = stats.ttest_1samp(our_column, meu, alternative = 'greater') # Greater means HO: <= and alternate : >
```

```python
print('T-Statistic :',result.statistic)
print('P-Value (One-Tailed) :',result.pvalue)
```

```python
alpha = 0.05
if result.pvalue < alpha :
    print('Reject Ho (mean is significatly grater than ',meu,')')
else:
    print('Do not reject HO')
```

# Excercise

Excercise : Use loan dataset. Run two-tailed and one-tailed one sample t-test\
Two-tailed -> HO : Average income = 50, H1 : Average income <> 50.\
One-tailed -> HO : Average income >= 50, H1 : Average income < 50.

```python
df = pd.read_csv('loan.csv')
df.head()
```

```python
our_column = df['income']
our_column = our_column.sample(n=30,replace=False)
```

```python
# Two Tailed
meu = 50
result = stats.ttest_1samp(our_column, meu)
print('T-Statistic :',result.statistic)
print('P-Value (Two-Tailed) :',result.pvalue)
```

```python
alpha = 0.05
if result.pvalue < alpha :
    print('Reject Ho')
else:
    print('Do not reject HO')
```

```python
# Two Tailed
result = stats.ttest_1samp(our_column, meu, alternative = 'less')
print('T-Statistic :',result.statistic)
print('P-Value (Two-Tailed) :',result.pvalue)
```

```python
alpha = 0.05
if result.pvalue < alpha :
    print('Reject Ho ( mean is significatly less than ',meu,')')
else:
    print('Do not reject HO')
```
