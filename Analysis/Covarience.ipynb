```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv(r'C:\Users\PGCP-AI.STUDENTSDC\Downloads\SP500 oil gold bitcoin.csv')
df
```

```python
corr = df[['BITCOIN', 'Gold', 'S&P500']]. corr()
cov = df[['BITCOIN', 'Gold', 'S&P500']]. cov()
print(corr)
print(cov)
print(cov.to_string(float_format = '{:.4f}'.format))
plt.figure(figsize=(10,6))
plt.scatter(df['BITCOIN'], df['Gold'], color = 'red', alpha = 0.5 , label = 'BITCOIN vs GOLD')
plt.scatter(df['BITCOIN'], df['S&P500'], color = 'green', alpha = 0.5 , label = 'BITCOIN s&p')
plt.scatter(df['S&P500'], df['Gold'], color = 'yellow', alpha = 0.5 , label = 's&p vs GOLD')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
df_2015 = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Downloads\marathon_results_2015.csv")
df_2016 = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Downloads\marathon_results_2016.csv")
df_2017 = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Downloads\marathon_results_2017.csv")
```

```python
df = pd.concat([df_2015, df_2016,df_2017], ignore_index=True)
sample_df = df.sample(frac=0.1, random_state=42)
print(df.head())
print(sample_df.head())
print(f'number of rows in the original dataframe:{len(df)}')
print(f'number of rows in the sampled dataframe:{len(sample_df)}')
```

```python
print(f'Average age - population:{df['Age'].mean():.2f}, sample: {sample_df['Age'].mean():.2f}')
```

```python
total_males = len(df[df['M/F'] == 'M'])
total_females = len(df[df['M/F'] == 'F'])
print(f'Total Male :{total_males} , Total Female :{total_females}')
```

```python
sample_males = int(0.1 * total_males)
sample_females = int(0.1 * total_females)
print(f'Sample Needs {sample_males} males and {sample_females} femals')
```

```python
male_sample = df[df['M/F'] == 'M'].sample(n=sample_males, random_state=42)
female_sample = df[df['M/F'] == 'M'].sample(n=sample_females, random_state=42)
sample_df = pd.concat([male_sample, female_sample])
```

```python
print('First Few rows of the sampled Dataframe :')
print(sample_df.head())
print(f'\nNumbers of rows in the original dataframe:{len(df)}')
print(f'Numers of row in sampled dataframe:{len(sample_df)}')
print(f'Numer of male athletes in the sampled dataframe:{(sample_df['M/F'] == 'M').sum()}')
print(f'Numer of female athletes in the sampled dataframe:{(sample_df['M/F'] == 'F').sum()}')
```

```python
# cluster sampling : clusters = countries
unique_countries = df['Country'].unique()
new_countries = len(unique_countries)

#Randomly select half of the countries
np.random.seed(42)
#np.random.choice picks random items from as array; replace = False -> no country picked twice
selected_countries = np.random.choice(unique_countries, size=new_countries // 2, replace=False)
print(selected_countries)
```

```python
#all records of the selected countries 
sample_df = df[df['Country'].isin(selected_countries)]
print('\nFirst few rows of the sampled dataframe:')
print(sample_df.head())
print(f'\nFirst few rows of the original dataframe:{len(df)}')
print(f'\nFirst few rows of the sampled dataframe:{len(sample_df)}')
```
