# Test 1: One-sample t-test

Suppose a company claims that its employees take an average of 30 minutes to complete a task.

```python
from scipy.stats import ttest_1samp
import numpy as np
```

```python
data = [32, 28, 35, 30, 29, 33, 31, 27, 34, 32]

# Define the hypotheses
# Null hypothesis (H0): The population mean is 30 minutes.
# Alternative hypothesis (Ha): The population mean is not 30 minutes.

meau = np.mean(data)
alpha = 0.05

result = ttest_1samp(data,popmean=30)

print('Sample Mean :-',meau)
print('t-statistic :-',result.statistic)
print('p-value :-',result.pvalue)

if result.pvalue < alpha :
    print('Null hypothesis (H0): The population mean is 30 minutes. -> rejected H0')
else :
    print("Fail to reject H0: Insufficient evidence to support Ha.")


# Define the hypotheses
# Null hypothesis (H0): The population mean is greater the >= 30 minutes.
# Alternative hypothesis (Ha): The population mean is not greater than 30 minutes.

result = ttest_1samp(data , popmean=30, alternative='less')

print('t-statistic :-',result.statistic)
print('p-value :-',result.pvalue)

if result.pvalue < alpha :
    print('Null hypothesis (H0): The population mean is greater than or equal to 30 minutes. -> rejected H0')
else :
    print("Fail to reject H0: Insufficient evidence to support Ha.")

# Define the hypotheses
# Null hypothesis (H0): The population mean is greater the <= 30 minutes.
# Alternative hypothesis (Ha): The population mean is not less than 30 minutes.

result = ttest_1samp(data , popmean=30, alternative='greater')

print('t-statistic :-',result.statistic)
print('p-value :-',result.pvalue)

if result.pvalue < alpha :
    print('Null hypothesis (H0): The population mean is less than or equal to 30 minutes. -> rejected H0')
else :
    print("Fail to reject H0: Insufficient evidence to support Ha.")
```

# Test 2: Independent Two-Sample t-test.

A company wants to determine whether two training methods produce different average test scores.\
Group A: Students trained using Method A.\
Group B: Students trained using Method B.

```python
from scipy.stats import ttest_ind


group_a = [72, 75, 78, 71, 74, 77, 73, 76]
group_b = [80, 82, 78, 85, 81, 79, 83, 84]

print('Mean Of Group A :-',np.mean(group_a))
print('Mean Of Group B :-',np.mean(group_b))

result = ttest_ind(group_a,group_b,alternative='two-sided')
alpha = 0.05

if result.pvalue < alpha:
    print('H0 : Both Group\'s Mean are Same.')
else:
    print('Failed to reject H0')

result = ttest_ind(group_a,group_b,alternative='greater')
alpha = 0.05

if result.pvalue < alpha:
    print("Reject H0: Evidence that Group A's mean is greater than Group B's.")
else:
    print("Fail to reject H0: Insufficient evidence that Group A's mean is greater than Group B's.")

result = ttest_ind(group_a,group_b,alternative='less')
alpha = 0.05

if result.pvalue < alpha:
    print("Reject H0: Evidence that Group A's mean is less than Group B's.")
else:
    print("Fail to reject H0: Insufficient evidence that Group A's mean is less than Group B's.")
```

# Test 3: Paired t-test.

A paired t-test compares two sets of measurements taken from the same people or related subjects.

```python
# H_0:  the population's mean difference is zero.
# H_a: the population's mean difference is not zero.

before = [60, 65, 70, 58, 62]
after  = [68, 70, 75, 65, 67]
alpha = 0.05
```

```python
from scipy.stats import ttest_rel
```

```python
result = ttest_rel(before,after,alternative='two-sided')
print('t-statistical :-',result.statistic)
print('p-value :-',result.pvalue)

if result.pvalue < alpha:
    print("Reject H0: Evidence that the population mean difference is not zero.")
else:
    print("Fail to reject H0: Insufficient evidence that the population mean difference is not zero.")
```

```python
# H_0:  the population's mean difference is greater than equal to zero.
# H_a: the population's mean difference is not greater zero.
#  H0 >= 0

result = ttest_rel(before,after,alternative='less')
print('t-statistical :-',result.statistic)
print('p-value :-',result.pvalue)


if result.pvalue < alpha:
    print("Reject H0: Evidence that before scores are lower than after scores on average.")
else:
    print("Fail to reject H0: Insufficient evidence that before scores are lower than after scores on average.")
```

```python
# H_0:  the population's mean difference is less than equal zero.
# H_a: the population's mean difference is not less zero.
#  H0 <= 0

result = ttest_rel(before,after,alternative='greater')
print('t-statistical :-',result.statistic)
print('p-value :-',result.pvalue)


if result.pvalue < alpha:
    print("Reject H0: Evidence that before scores are greater than after scores on average.")
else:
    print("Fail to reject H0: Insufficient evidence that before scores are greater than after scores on average.")
```

# Test 4: Z-test

```python
# 1. Calculate the Z-statistic using the formula.
# 2. Calculate the two-sided p-value using norm.cdf().
# 3. Compare the p-value with alpha.
# 4. Decide whether to reject \(H_0\).
# 5. Write your conclusion in simple words.
# Hypotheses:
# H0 : mue = 50
# Ha : mue != 50
```

```python
from scipy.stats import norm
sample_mean = 53
population_mean = 50
population_std = 10
n = 36
alpha = 0.05

z_stat = (sample_mean - population_mean)/(population_std/np.sqrt(n))
print('Z-Statistic :-',z_stat)

p_value = 2 * (1 - norm.cdf(z_stat))
print('P-Value :-',p_value)

if p_value < alpha :
    print('Reject H0 :- Evidance that Population mean doesn not equalt to 50')
else :
    print('Does not reject H0 :- Insufficient Evidence thet Population mean does not equal to 50')
```

```python
# Right-tailed Z-test
# Use this when you want to test whether the population mean is greater than 50.
# H0 : mue <= 50
# Ha : mue > 50

z_stat = (sample_mean - population_mean)/(population_std/np.sqrt(n))
print('Z-Statistic :-',z_stat)

p_value = 1 - norm.cdf(z_stat)
print('P-Value :-',p_value)

if p_value < alpha:
    print("Reject H0: Evidence that the population mean is greater than 50.")
else:
    print("Fail to reject H0: Insufficient evidence that the population mean is greater than 50.")
```

```python
# Right-tailed Z-test
# Use this when you want to test whether the population mean is greater than 50.
# H0 : mue >= 50
# Ha : mue < 50

z_stat = (sample_mean - population_mean)/(population_std/np.sqrt(n))
print('Z-Statistic :-',z_stat)

p_value = norm.cdf(z_stat)
print('P-Value :-',p_value)

if p_value < alpha :
    print('Reject H0 : Evidence that population mean less than  5o')
else :
    print('failed to Reject H0 : Insufficient Evidence that population mean less than 5o')
```

# Test 5: Chi-square test.

A Chi-square test helps us analyze categorical data—data divided into groups or categories.

```python
# H0: Gender and product preference are independent (no association).
# Ha: Gender and product preference are associated.

observed = [
    [40, 20],
    [25, 35]
]

from scipy.stats import chi2_contingency
result = chi2_contingency(observed)
print('Chi Square Statistic :-',result.statistic)
print('P-Value :-',result.pvalue)
print('Degree of Freedom :-',result.dof)
print('Excepcted Values :-',result.expected_freq)
```

```python
# H0: All four categories are equally popular.
# Ha: The category proportions are not all equal.

observed = [18, 22, 27, 33]

# Q. if the total count is 100, how many observations would we expect in each category under H0? 
# -> 100/4 -> 25 expected in each
from scipy.stats import chisquare

result = chisquare(observed,f_exp=[25,25,25,25])
print('Chi Square Statistic :-',result.statistic)
print('P-Value :-',result.pvalue)

alpha = 0.05

if result.pvalue < alpha:
    print("Reject H0: Evidence that category proportions are not all equal.")
else:
    print("Fail to reject H0: Insufficient evidence that category proportions are not all equal.")
```

# Test 6: One-Way ANOVA test.
ANOVA (Analysis of Variance) is a statistical test used to compare the means of three or more independent groups.

```python
group_a = [70, 72, 68, 74, 71]
group_b = [80, 78, 85, 82, 81]
group_c = [65, 67, 64, 69, 66]

print('Mean of A :-',np.mean(group_a))
print('Mean of B :-',np.mean(group_b))
print('Mean of C :-',np.mean(group_c))

alpha = 0.05

# H0: All three population means are equal.
# Ha: At least one population mean is different.

from scipy.stats import f_oneway

result = f_oneway(group_a,group_b,group_c)
print('F-statistic :-', result.statistic)
print('p-Value :-',result.pvalue)

if result.pvalue < alpha :
    print('Reject H0 : Evidence that all population mean are not equal')
else :
    print('Failed to Reject H0 : Insufficient Evidence that all population are not equal')

if result.pvalue <= alpha:
    print("Reject H0: Evidence that at least one population mean differs.")
else:
    print("Fail to reject H0: Insufficient evidence that at least one population mean differs.")
```

# Test 7 : Mann–Whitney U Test.
The Mann–Whitney U test is a non-parametric test used to compare two independent groups.

```python
# H0: The two groups have the same distribution.
# Ha: The two groups have different distributions.

group_a = [12, 15, 14, 10, 18, 16, 13]
group_b = [22, 25, 19, 24, 21, 27, 23]

alpha = 0.05

from scipy.stats import mannwhitneyu

result = mannwhitneyu(group_a,group_b,alternative='two-sided')

print('U-statistic:', result.statistic)
print('P-value :-',result.pvalue)

if result.pvalue < alpha:
    print("Reject H0: Evidence that the groups have different distributions.")
else:
    print("Fail to reject H0: Insufficient evidence that the groups have different distributions.")
```
