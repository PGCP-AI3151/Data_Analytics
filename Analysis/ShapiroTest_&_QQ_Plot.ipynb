```python
# check if data is mormally distributed : shapiro test and qq plot
```

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import scipy.stats as stats
import seaborn as sns
```

```python
data = sns.load_dataset('tips')
total_bill_series = data['total_bill']
```

```python
# Normality Test with Shapiro-Wilk test
shapiro_test = stats.shapiro(total_bill_series)
print('Test statistics (w): ' , shapiro_test.statistic)
print('p-value:' , shapiro_test.pvalue)
```

```python
alpha = 0.05 #common significance level
if shapiro_test.pvalue > alpha:
    print('Data is normally distributed')
else:
    print('Data is not normally distributed')
    
```

```python
# QQ Plot
stats.probplot(total_bill_series, dist='norm', plot=plt)
plt.title('QQ Plot Of Total Bill')
plt.xlabel('Theoratical Quantiles (Normal Distribution)')
plt.ylabel('Ordered Values (Total Bill)')
plt.grid(True)
plt.show()
```
