```python
import numpy as np
import pandas as pd

np.random.seed(42)

df = pd.DataFrame({
    "normal": np.random.normal(50, 10, 1000),
    "right_skewed": np.random.exponential(2, 1000),
    "left_skewed": -np.random.exponential(2, 1000) + 10,
})

df.head()
```

```python
df['normal'].skew()
```

```python
stats.probplot(df['normal'],dist='norm',plot=plt)
plt.show()
```

```python
df['right_skewed'].skew()
```

```python
plt.hist(df['right_skewed'])
```

```python
stats.probplot(df['right_skewed'],dist='norm',plot=plt)
plt.show()
```

```python
df['left_skewed'].skew()
```

```python
plt.hist(df['left_skewed'])
```

```python
stats.probplot(df['left_skewed'],dist='norm',plot=plt)
plt.show()
```

```python
# Right-skewed → upper/right tail tends to bend upward ↗️
# Left-skewed → lower/left tail tends to bend downward ↙️
```

```python
from scipy.stats import shapiro

stat,p_val = shapiro(df['normal'])
print(stat)
print(p_val)
```

```python
alpha = 0.05
if p_val > alpha:
    print('Data is approximately normally distributed')
else:
    print('Data is not normally distributed')
```
