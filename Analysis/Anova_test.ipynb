```python
# ANOVA : HO: Partent's education level has no impact on the child's math score
```

```python
from scipy.stats import f_oneway
import pandas as pd
```

```python
df = pd.read_csv('StudentsPerformance.csv')
df.head()
```

```python
g1 = df['math score'][df['parental level of education'] == 'some high school']
g2 = df['math score'][df['parental level of education'] == 'high school']
g3 = df['math score'][df['parental level of education'] == 'some college']
g4 = df['math score'][df['parental level of education'] == 'associate\'s degree']
g5 = df['math score'][df['parental level of education'] == 'bachelor\'s degree']
g6 = df['math score'][df['parental level of education'] == 'master\'s degree']
```

```python
fstat, pval = f_oneway(g1,g2,g3,g4,g5,g6)
alpha = 0.05

if pval < alpha :
    print('Reject HO')
else :
    print('Does Not Reject HO')
```

```python
print("g1 size:", len(g1))
print("g2 size:", len(g2))
print("g3 size:", len(g3))
print("g4 size:", len(g4))
print("g5 size:", len(g5))
print("g6 size:", len(g6))
```

# Excercise

Excercise : Is there a statistically significant difference between how people tip, whose bill amounts are < $13, $13-17,$17-22,>$22 in the tips dataset? (Handle Boundry Condition Appropriately)

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Downloads\tips.csv")
df.head()
```

```python
g1 = df['tip'][df['total_bill'] < 13]
g2 = df['tip'][df['total_bill'].between(13,17,inclusive='left')]
g3 = df['tip'][df['total_bill'].between(17,22,inclusive='left')]
g4 = df['tip'][df['total_bill'] >= 22]
```

```python
fstat, pval = f_oneway(g1,g2,g3,g4)

print('F-Statistic :',fstat)
print('P-Value :',pval)

alpha = 0.05

if pval < alpha :
    print('Reject HO')
else :
    print('Does Not Reject HO')
```

```python
print("g1 size:", len(g1))
print("g2 size:", len(g2))
print("g3 size:", len(g3))
print("g4 size:", len(g4))
```
