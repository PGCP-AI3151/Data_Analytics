```python
from scipy.stats import binom
```

```python
n = 5 # no. of students asked (trials)
k= 3 # no. who like python(success)
p = 0.66 # prob that one student likes python
```

```python
pmf = binom.pmf(k,n,p) # pmf:p(x=k)
cdf = binom.cdf(k, n, p) # cdf: p(x<=k)
print(f"PMF (p(x = {k})) = {pmf:.4F}")
print(f"CDF (p(x <= {k})) = {cdf:.4F}")
print(binom.pmf(3, 5, 0.66)) # acts like pdf
```

```python
#pdf
prob = binom.cdf(3,n,p) - binom.cdf(1,n,p)
print(f'Bonomial Probability P(2 <= X <= 3) = {prob :.4f}')

#pdf - another method
pmf_2 = binom.pmf(2,n,p)
pmf_3 = binom.pmf(3,n,p)
prob = pmf_2 + pmf_3
print(f'Binomial Probability P(2 <= x <= 3) = {prob :.4f}')
```

```python
""" Excercise : 75% of patient with a disease die of it. six paatients are selected. 
 What is the probability that (a) exactly 4 recover, (b) more than 4 recover ? """
```

```python
n , p = 6, 0.25 #SUCCESS == RECOVER, so p = 0.25
print("P(Exactly 4 Recover) =",round(binom.pmf(4,n,p),4))  
print("P(More Than 4 Recover) =",round(1 - binom.cdf(4,n,p),4)) # = pmf(5) + pmf(6)
```

```python
"""A marksman hits the target 4 times out of 5. find probability that when he fires 4 shorts:  
(a) more than 2 are hits (b) at least 3 are misses (= at most 1 hit) """
```

```python
n ,p = 4, 0.80
print(f'More than 2 hits =',1 - round(binom.cdf(2,n,p),4))
print(f'At Least 3 are misses =',round(binom.cdf(1,n,p),4))

# Solution
print(f'More than 2 hits =',binom.pmf(3,n,p) + binom.pmf(4,n,p))
print(f'At Least 3 are misses =',binom.pmf(0,n,p) + binom.pmf(1,n,p))
```

```python
""" 65% of students who appear for the c_DAC entrance test clear it. from a city, 50 students appear. 
find probability that (a) nobody cleaqars, (b) more than 40 clear """
```

```python
""" In a call center , 20% of calls need a supervisor follow up. 50 calls have come this morning. find probability that
(a) no calls need follow up
(b) more than 15 calls need follow up
(c) less than 10 calls need follow up  """
```

```python
n , p = 50 , 0.20
print("no calls need follow up =",binom.pmf(0,n,p))
print("more than 15 calls need follow up =",1 - binom.cdf(15,n,p))
print("less than 10 calls need follow up =",binom.cdf(9,n,p))
```
