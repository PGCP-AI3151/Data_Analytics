```python
# Chi-Square test of independence
# HO : There is NO Statistically significant dependence between Cancelled/Pending orders and Order Fulfilment.
```

```python
from scipy.stats import chi2_contingency
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Downloads\Amazon Sale Report.csv",low_memory=False,index_col=0)
df.head()
```

```python
cancelled_pending_orders = df[df['Status'].isin(['Cancelled','Pending'])]
```

```python
# Create a Contingency table
contingency_table = pd.crosstab(cancelled_pending_orders['Status'],cancelled_pending_orders['Fulfilment'])
contingency_table
```

```python
# Perofrm chi-square test
chi2,p,dof,ex = chi2_contingency(contingency_table)
print('Chi-Squared Statistic :',chi2)
print('P-Value :',p)
print('Degrees of Freedom :',dof)
print('Expected Values :',ex)
```

```python
if p < 0.05 :
    print('Reject Null Hypothesis.')
else :
    print('Fail to Reject Null Hypothesis.')
```

# Excercise

Excercise : \
(a) Were gender and survival independent in titanic?\
(b) Were Passanger class and survival related in titanic?

```python
import seaborn as sns

df = sns.load_dataset('titanic')
df.head()
```

```python
genders = df[df['sex'].isin(['male','female'])]
```

```python
contingency_table = pd.crosstab(genders['sex'],genders['survived'])
contingency_table
```

```python
# Perofrm chi-square test
chi2,p,dof,ex = chi2_contingency(contingency_table)
print('Chi-Squared Statistic :',chi2)
print('P-Value :',p)
print('Degrees of Freedom :',dof)
print('Expected Values :',ex)
```

```python
if p < 0.05 :
    print('Reject Null Hypothesis.')
else :
    print('Fail to Reject Null Hypothesis.')
```
