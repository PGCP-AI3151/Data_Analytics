```python
import pandas as pd
from scipy import stats
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Data_Analytics\DA_Pandas\tips.csv")
df.head()
```

```python
male_tips = df[df['gender'] == 'Male']['tip']
female_tips = df[df['gender'] == 'Female']['tip']
```

```python
male_tips = male_tips.sample(n=30,replace=False)
female_tips = female_tips.sample(n=30,replace=False)

print('Male mean tip :',male_tips.mean())
print('Female mean tip :',female_tips.mean())
```

```python
# perform two_sample t-test
t_statistic, p_value = stats.ttest_ind(male_tips, female_tips)
```

```python
# Significance level (alpha)
alpha = 0.05
print('T-Statistic :',t_statistic)
print('P-Value :',p_value)
```

```python
if p_value < alpha :
    print('There is a significant fifference in spending score between Male and Female groups.')
else:
    print('There is a no significant fifference in spending score between Male and Female groups.')    
```

# Excercise

Excercise : Is there a statistically significant difference in the average runs scored by Sachin Tendulkar when playing against Australia versus Pakistan ? (sachin_tendulkar.csv)

```python
df = pd.read_csv('sachin_tendulkar.csv')
df.head()
```

```python
vs_aus = df[df['Versus'] == 'Australia']['Runs']
vs_pak = df[df['Versus'] == 'Pakistan']['Runs']
```

```python
vs_aus = vs_aus.sample(n=30,replace=False)
vs_pak = vs_pak.sample(n=30,replace=False)

print('Vs Australia Mean Runs :',vs_aus.mean())
print('Vs Pakistan Mean Runs :',vs_pak.mean())
```

```python
result = stats.ttest_ind(vs_aus,vs_pak)
print('T-Statistic :',result.statistic)
print('P-Value :',result.pvalue)
```

```python
alpha = 0.05
if p_value < alpha :
    print('There is a Significant Difference in Mean Runs score by Sacin Tendulkar Versus Australia and Pakistan.')
else:
    print('There is a No Significant Difference in Mean Runs score by Sacin Tendulkar Versus Australia and Pakistan.')   
```
