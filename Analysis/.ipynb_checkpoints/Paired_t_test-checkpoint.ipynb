```python
# Paired t_test : A company has launched some offer
# HO : Pre and post usage is the same (i.e offer is not effective)
```

```python
from scipy.stats import ttest_rel
import pandas as pd
```

```python
df = pd.read_csv('cust_seg.csv')
df.head()
```

```python
usage_data = df[['pre_usage','Post_usage_1month']]
usage_data = usage_data.sample(n=30)
```

```python
tstat, pval = ttest_rel(usage_data['pre_usage'],usage_data['Post_usage_1month'])

alpha = 0.05
if pval < alpha :
    print('Reject null hypothesis')
else :
    print('Fail top reject null hypothesis')
```

# Excercise

Excercise : On the Hypothermia dataset , check whether the urgent treatment is effective (consider column t.1 and t.2)

```python
df = pd.read_csv('Hypothermia.csv')
df.head()
```

```python
use_data = df[['t.1','t.2']]
use_data = use_data.sample(n=30)
```

```python
print('Use Data Mean :',use_data.mean())
```

```python
tstat, pval = ttest_rel(use_data['t.1'],use_data['t.2'])

alpha = 0.05
if pval < alpha :
    print('Reject null hypothesis')
else :
    print('Fail top reject null hypothesis')
```
