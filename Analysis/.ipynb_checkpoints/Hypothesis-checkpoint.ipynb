```python
# smaple standard deviation = 10 ml
# test at 5% significfance level wheteher the average is greater than 500 ml
```

```python
from scipy.stats import norm
mu = 500
x_bar = 503.4
s = 10
n = 25
alpha = 0.05
```

```python
#standard error
se = s/n**0.5
```

```python
#z test statistics
z = (x_bar - mu) / se
```

```python
# critical value 
#percent point func prob - zscore
#if probability = 0.025, z_score = 1.96
critical_value = norm.ppf(1 - alpha/2)
```

```python
from scipy.stats import norm
#p value
#survival func - gives the prob to the right of a zscore 
p_value = 2*norm.sf(abs(z))
print('Population mean =', mu)
print('sample mean = ', x_bar)
print('std deviation =', s)
print('sample size =', n)
print('alpha =', alpha)
print('std error =', se)
print('z stats =', z)
print('critical_value =',critical_value)
print('p value =', p_value)
```

```python
#critical value method
if z > critical_value:
    print('critical value method: Reject HO')
else:
     print('critical value method: Do not reject HO')
    
```

```python
# p value method
if p_value < alpha:
     print('P-value method: Reject HO')
else:
    print('P-value method: Do not reject HO')
```
