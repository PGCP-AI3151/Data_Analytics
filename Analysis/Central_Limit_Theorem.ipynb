```python
import pandas as pd 
import matplotlib.pyplot as plt
import scipy.stats as stats
import numpy as np
```

```python
df = pd.read_csv('diabetes.csv')
df.head()
```

```python
# Plot The Distribution of Glucose
plt.hist(df['Glucose'], bins=30, edgecolor='black', alpha=0.7)
plt.title('Distribution of Sample Means (n=30)')
plt.xlabel('Sample Mean')
plt.ylabel('Frequency')
plt.show()
```

```python
# Generate a Q-Q plot for the Glucose
stats.probplot(df['Glucose'], dist='norm',plot=plt)
plt.title('Q-Q Plot of Sample Means')
plt.show()
```

```python
# perform the Shapiro-wilk test on the Glucose
shapiro_test = stats.shapiro(df['Glucose'])
print('Shapiro-Wilk p-Value:',shapiro_test.pvalue)
```

```python
# Interpret the result
if shapiro_test.pvalue < 0.05:
    print('Likely does not follow a normal distribution.')
else:
    print('May be normally distributed.')
```

```python
# Central Limit Theorem
sample_means = []
n_sample = 100
sample_size = 30

for i in range(n_sample):
    sample = df['Glucose'].sample(n=sample_size, replace=True)
    sample_means.append(sample.mean())
```

```python
# Plot The Distribution of sample means
plt.hist(sample_means, bins=30, edgecolor='black', alpha=0.7)
plt.title('Distribution of Sample Means (n=30)')
plt.xlabel('Sample Mean')
plt.ylabel('Frequency')
plt.show()
```

```python
# Generate a Q-Q plot for the sample means
stats.probplot(sample_means, dist='norm',plot=plt)
plt.title('Q-Q Plot of Sample Means')
plt.show()
```

```python
# perform the Shapiro-wilk test on the sample means
shapiro_test = stats.shapiro(sample_means)
print('Shapiro-Wilk p-Value:',shapiro_test.pvalue)
```

```python
# Interpret the result
if shapiro_test.pvalue < 0.05:
    print('Likely does not follow a normal distribution.')
else:
    print('May be normally distributed.')
```

```python
population_size = 100000
sample_size = 50
samples = 2000

population = np.random.exponential(scale=2,size=population_size)
```

```python
# perform the Shapiro-wilk test on the sample means
shapiro_test = stats.shapiro(population)
print('Shapiro-Wilk p-Value:',shapiro_test.pvalue)
```

```python
# Interpret the result
if shapiro_test.pvalue < 0.05:
    print('Likely does not follow a normal distribution.')
else:
    print('May be normally distributed.')
```

```python
stats.probplot(population,dist='norm',plot=plt)
plt.show()
```

```python
sample_mean = []
for i in range(samples):
    sample = np.random.choice(population,size=sample_size)
    sample_mean.append(sample.mean())
```

```python
stats.probplot(sample_mean,dist='norm',plot=plt)
plt.show()
```

```python
shapiro_test = stats.shapiro(sample_means)
print('Shapiro-Wilk p-Value:',shapiro_test.pvalue)
```

```python
# Interpret the result
if shapiro_test.pvalue < 0.05:
    print('Likely does not follow a normal distribution.')
else:
    print('May be normally distributed.')
```
