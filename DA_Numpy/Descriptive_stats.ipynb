```python
import numpy as np
sales = np.random.randint(100,300,15)
print(sales)
```

```python
print("Total :", sales.sum())
print("Mean :", sales.mean())
print("Median :", np.median(sales))
print("Min :", sales.min(), 'Max:' , sales.max())
print("Range :", np.ptp(sales))
print("Var :", sales.var().round(2))
print("Std :", sales.std().round(2))
q1,q2,q3 = np.percentile(sales,[25,50,75])
print("Q1 Q2 Q3 :",q1,q2,q3,"IQR :",q3-q1)
```

```python
# With Specific Axis
store_sales = np.random.randint(100,300,(3,4))
print(store_sales)
```

```python
print(store_sales.sum())#sums everything
print(store_sales.sum(axis=0))#rows sum
print(store_sales.sum(axis=1))#col sum
```
