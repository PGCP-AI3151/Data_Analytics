```python
# Where 
```

```python
result_array = np.where(sales_array == 0, 'No Sales','Some Sales')
print(result_array)
```

```python
# boolean mask 
bool_mask = sales_array != 0
print(bool_mask)
```

```python
print(sales_array[bool_mask])
```
