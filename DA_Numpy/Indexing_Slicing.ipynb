```python
import numpy as np
np.random.seed(1)
marks = np.random.randint(30,101,10)
print(marks)
```

```python
print(marks[8])
```

```python
print(marks[-1])
```

```python
print(marks[1:5])
```

```python
print(marks[:3])
```

```python
print(marks[-3:])
```

```python
print(marks[::2])
```

```python
print(marks[::-1])
```

```python
# Assign one value to whole slice
```

```python
marks[0:3] = 75
print(marks)
```

```python
# Shallow Copy
```

```python
top_batch = marks[0:5]
top_batch[:] = 99
print(top_batch) #original change
print(marks)
```

```python
# Deep Copy
```

```python
marks_copy = marks.copy()
marks_copy[:] = 99
print(marks_copy) #original unchanged
print(marks)
```

```python
#conitional (boolean) selection
```

```python
marks = np.array([23,45,65,76,87,25])
passed = marks >=35
print(passed)
print(marks[passed])
print(marks[marks > 70])
```

#multiple condition: use & , | , ~ wityh brackets around each condition

```python
print(marks[ (marks >= 50) & (marks < 80)])
print(marks[ (marks < 35) & (marks > 90)])
print(marks[~passed])
print(np.sum(marks < 35))
```
