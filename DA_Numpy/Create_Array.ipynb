```python
# Random Numbers  -> every run got different random number
try1  = np.random.randint(0,10,5)
print(try1)
```

```python
# with seed -> every run you got same random number (value of seed -> any integer but standard 1,42,101)
np.random.seed(42)
try1  = np.random.randint(0,10,5)
print(try1)
```
