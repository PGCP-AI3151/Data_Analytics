```python
import numpy as np
```

```python
marks = np.array([78,95,88,67,91])
names = np.array(['Amit','Neha','Raj','Priya','Kiran'])
```

```python
print(np.sort(marks))  # ascending by default
```

```python
print(np.sort(marks)[::-1])  # descending by reverse ascending
```

```python
order = np.argsort(-marks)  # -ve -> for descending
print(order)
print(names[order])
print(marks[order])
```

```python
order = np.argsort(marks)  # +ve -> for ascending
print(order)
print(names[order])
print(marks[order])
```

```python
expenses = np.array([2500,3200,np.nan,4100,3900,np.nan,3000])
print(expenses)
```

```python
print(expenses.sum())
print(expenses.mean())
```

```python
print(np.nan == np.nan) # -> nan is not equal to anything, not even to itself
```

```python
# finding missing values
```

```python
missing = np.isnan(expenses)
print(missing)
```

```python
print('Missing Count :',missing.sum())
```

```python
# Drop : keep only non-missing values
clean = expenses[~np.isnan(expenses)]
print(clean)
```

```python
# Filled With Fixed Value
print(np.nan_to_num(expenses,nan=0))
```

```python
# Filled With the mean (mean imputation)
filled = np.where(np.isnan(expenses),np.nanmean(expenses),expenses)
print(filled.round(2))
```
